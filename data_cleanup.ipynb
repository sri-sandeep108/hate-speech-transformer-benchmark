```python
import pandas as pd
from tqdm import tqdm
```

```python
df = pd.read_csv("hate_speech_clean_master.csv")
print(df.head())
```

```python
from transformers import AutoTokenizer

print(f"Original Row Count: {len(df)}")
tokenizer = AutoTokenizer.from_pretrained("distilbert-base-uncased")
```

```python
df['token_len'].value_counts()
```

```python
def get_token_count(text):
    tokens = tokenizer.encode(str(text), add_special_tokens=True)
    return len(tokens)
```

```python
print("Counting tokens...")
df["token_len"] = [
    get_token_count(text) for text in tqdm(df["text"], desc="Processing tokens")
]
```

```python
clean_df = df[df["token_len"] <= 500]
```

```python
dropped_count = len(df) - len(clean_df)
print(f"Dropped {dropped_count} rows that were too long.")
print(f"Final Row Count: {len(clean_df)}")
```

```python
clean_df.to_csv("hate_speech_clean_master.csv", index=False)
```
