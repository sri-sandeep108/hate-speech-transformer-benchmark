```python
import spacy
from IPython.display import display, HTML, clear_output
import pandas as pd
from tqdm import tqdm
```

```python
MODEL_PATHS = {
    "Static": "output/static_vector/model-best",
    "BERT": "output/bert/model-best",
    "RoBERTa": "output/roberta/model-best",
    "DistilBERT": "output/distilbert/model-best",
    "ELECTRA": "output/electra/model-best",
}

nlp_models = {name: spacy.load(path) for name, path in MODEL_PATHS.items()}
```

```python
def predict_all_models(text, nlp_models):
    results = []

    for model_name, nlp in nlp_models.items():
        doc = nlp(text)
        hateful_score = doc.cats.get("Hateful", 0)
        not_hateful_score = doc.cats.get("Not-Hateful", 0)

        prediction = "Hateful" if hateful_score > not_hateful_score else "Not-Hateful"

        results.append(
            {
                "Model": model_name,
                "Prediction": prediction,
                "Hateful Score": round(hateful_score, 4),
                "Not-Hateful Score": round(not_hateful_score, 4),
            }
        )

    return pd.DataFrame(results)
```

```python
def display_results(text, df):
    html = f"""
    <div style="font-family: Arial; padding: 15px;">
        <h2>📝 Input Text</h2>
        <blockquote style="background:#f4f4f4; padding:10px; border-left:5px solid #888;">
            {text}
        </blockquote>

        <h2>📊 Model Predictions</h2>
        {df.to_html(index=False, escape=False)}
    </div>
    """
    display(HTML(html))
```

```python
text = input("\nEnter text (or type 'exit' to stop): ")

df_results = predict_all_models(text, nlp_models)
display_results(text, df_results)
```
